```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
L1 = np.array([-0.9,0,-3.1,3.7,1.7,-2])/100
L2 = np.array([-1.6,-0.8,-1.4,3.7,2.3,-2])/100
L = np.stack([L1,L2])
w = np.array([800,200])
```

```python
m1 = np.mean(L1)
m2 = np.mean(L2)
m = np.array([m1,m2])
S = np.cov(L1,L2,ddof=1)
print(m)
print(S)
```

```python
mu = -np.transpose(w)@m
Sigma = np.transpose(w)@S@w
print(mu)
print(Sigma)
```

```python
N1 = np.random.normal(loc=mu,scale=np.sqrt(Sigma),size=10000)
```

```python
-w@L
```

```python
L_t = np.sort((-w@L))
print(L_t)
y = np.arange(1, len(L_t) + 1) / len(L_t)

plt.ecdf(L_t,label="Historical simulation")
plt.ecdf(N1,label="Covar-Var")
plt.legend()
plt.show()
```
