```r
dd <- read.table("http://stat.ethz.ch/Teaching/Datasets/WBL/rainDay.txt", header=T)
```

```r
dd
```

```r
length(dd[["rain"]])
```

```r
data <-  ts(dd[["rain"]],frequency = 365,deltat = 1/365,start =c(2000,1))
```

```r
data_snip <- window(x = data,start = c(2005,3),end=c(2007,365))
```

```r
print(data_snip)
```
